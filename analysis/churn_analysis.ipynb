```python
import pandas as pd
import numpy as np
from datetime import timedelta
```

```python
customer_profiles = pd.read_parquet('F:/iti/python/project/NexaBank45/tmp/2025-05-14/20/customer_profiles.parquet')
customer_profiles
```

```python
transactions = pd.read_parquet('F:/iti/python/project/NexaBank45/tmp/2025-05-14/20/transactions.parquet')
transactions
```

```python
credit_cards_billing = pd.read_parquet('F:/iti/python/project/NexaBank45/tmp/2025-05-14/20/credit_cards_billing.parquet')
credit_cards_billing
```

```python
transactions['transaction_date'] = pd.to_datetime(transactions['transaction_date']).dt.date
analysis_date = pd.to_datetime('today').date()
last_transaction = transactions.groupby('sender')['transaction_date'].max()
cutoff_date = analysis_date - timedelta(days=90)
churned_customers = last_transaction[last_transaction < cutoff_date].index

print(f"Total customers: {len(last_transaction)}")
print(f"Churned customers: {len(churned_customers)}")
print(f"Percentage churned: {len(churned_customers)/len(last_transaction):.1%}")
```

```python
customer_profiles['age_group'] = pd.cut(customer_profiles['age'],
                                       bins=[0, 30, 40, 50, 60, 100],
                                       labels=['<30', '30-39', '40-49', '50-59', '60+'])
customer_profiles
```

```python
customer_spending = transactions.groupby('sender')['transaction_amount'].sum()
customer_profiles = customer_profiles.merge(
    customer_spending.rename('total_spending'),
    left_on='customer_id',
    right_index=True,
    how='left'
)
customer_profiles['spending_level'] = pd.qcut(
    customer_profiles['total_spending'],
    4,
    labels=['Low', 'Medium', 'High', 'VIP']
)
```

```python
customer_profiles['is_churned'] = customer_profiles['customer_id'].isin(churned_customers)

churn_by_age = customer_profiles.groupby('age_group')['is_churned'].mean()
churn_by_age
```

```python
churn_by_city = customer_profiles.groupby('city')['is_churned'].mean().sort_values(ascending=False)
churn_by_city
```

```python
churn_by_spending = customer_profiles.groupby('spending_level')['is_churned'].mean()
churn_by_spending
```

```python
credit_cards_billing['payment_date'] = pd.to_datetime(credit_cards_billing['payment_date'])
credit_cards_billing['month'] = pd.to_datetime(credit_cards_billing['month'])
credit_cards_billing['days_late'] = (credit_cards_billing['payment_date'] - credit_cards_billing['month']).dt.days - 15
```

```python
payment_behavior = credit_cards_billing.groupby('customer_id').agg(
    avg_days_late=('days_late', 'mean')  # Only compute average days late
)
payment_behavior
```

```python

# Merge with churn data
payment_churn = customer_profiles[['customer_id', 'is_churned']].merge(
    payment_behavior,
    on='customer_id',
    how='left'
)
payment_churn
```

```python
# Calculate correlations
# If 'is_churned' is boolean, convert to 0/1 first
payment_churn['is_churned'] = payment_churn['is_churned'].astype(int)
correlation = payment_churn[['is_churned', 'avg_days_late']].corr()['is_churned'][['avg_days_late']]
correlation
```
